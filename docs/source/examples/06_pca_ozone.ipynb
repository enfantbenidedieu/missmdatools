## ozone dataset

```python
# ozone dataset
from missmdatools.datasets import ozone
print(ozone.info())
```

```python
# display
print(ozone)
```

## Estimation of number of components - PCAestimncp

```python
# instanciation & training
from missmdatools import PCAestimncp
clf = PCAestimncp(sup_var=(11,12))
clf.fit(ozone)
```

```python
# optimal number of components
ncp = clf.ncp_
print(f"Optimal number of components = {ncp}")
```

## Missing values imputation - PCAimputer

### EM

```python
# instanciation & training
from missmdatools import PCAimputer
clf = PCAimputer(ncp=ncp,sup_var=(11,12))
clf.fit(ozone)
```

#### Complete dataset

```python
# complete dataset
print(clf.complete_)
```

#### Reconstructed data

```python
# reconstructed dataset
print(clf.fitted_)
```

### Iterative regularized algorithm

```python
# instanciation & training
clf = PCAimputer(ncp=ncp,method="regularized",sup_var=(11,12))
clf.fit(ozone)
```

#### Complete dataset

```python
# complete dataset
print(clf.complete_)
```

#### Reconstructed data

```python
# reconstructed dataset
print(clf.fitted_)
```

## ozoneNA dataset

```python
import pandas as pd
ozoneNA = pd.read_csv("http://factominer.free.fr/missMDA/ozoneNA.csv",header=0,sep=",",index_col=0)
print(ozoneNA.info())
```

```python
# display
print(ozoneNA)
```

### Estimation of number of components

```python
# instanciation & training
clf = PCAestimncp(cv="gcv",sup_var=11)
clf.fit(ozoneNA)
```

```python
# optimal number of components
ncp = clf.ncp_
print(f"Optimal number of components = {ncp}")
```

### Missing values imputation

```python
# instanciation & training
clf = PCAimputer(ncp=ncp,method="regularized",sup_var=11)
clf.fit(ozoneNA)
```

#### Complete dataset

```python
# complete dataset
print(clf.complete_)
```

#### Reconstructed dataset

```python
# reconstructed dataset
print(clf.fitted_)
```
